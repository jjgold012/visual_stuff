# Import Required Libraries
This section imports the necessary Python libraries for cellular automata simulation and visualization.

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation

```

# Define Cellular Automaton Rules
Here, we define the update rules for the cellular automaton. We'll use Conway's Game of Life as an example, but you can modify the rules for other automata.

```python
# Define Elementary Cellular Automaton Rule
# filepath: untitled:Untitled-1.ipynb?jupyter-notebook

def get_rule_table(rule_number):
    """Return the rule table for a given elementary CA rule (0-255)."""
    rule_bin = np.array([int(x) for x in np.binary_repr(rule_number, width=8)])
    # Map 3-bit neighborhood to rule output
    table = {}
    for i in range(8):
        key = tuple([int(x) for x in np.binary_repr(7-i, width=3)])
        table[key] = rule_bin[i]
    return table

def elementary_step(row, rule_table):
    """Compute next row for elementary CA."""
    n = len(row)
    new_row = np.zeros_like(row)
    for i in range(n):
        new_row[i] = rule_table[row[(i-1) % n], row[i], row[(i+1) % n]]
    return new_row
```

# Initialize the Grid
Set up the initial state of the cellular automaton grid. You can randomize the grid or specify a custom pattern.

```python
# Initialize 1D Grid
# filepath: untitled:Untitled-1.ipynb?jupyter-notebook

cells = 800
steps = 1600
n = 30
r = False
```

```python
import ipywidgets as widgets
from IPython.display import display

grid = np.zeros((steps, cells), dtype=int)
def run_and_plot(rule_number, random_start=False):
    rule_table = get_rule_table(rule_number)
    n = rule_number
    r = random_start
    if random_start:
        grid[0] = np.random.randint(0, 2, size=cells)
    else:
        grid[0, cells // 2] = 1
    for t in range(1, steps):
        grid[t] = elementary_step(grid[t-1], rule_table)
    plt.figure(figsize=(10,6))
    plt.imshow(grid, cmap='binary', interpolation='nearest')
    plt.title(f"Elementary Cellular Automaton - Rule {rule_number}")
    plt.xlabel("Cell")
    plt.ylabel("Time Step")
    plt.show()

widgets.interact(run_and_plot, rule_number=(0,255,1), random_start=widgets.Checkbox(value=False, description='Random Start'))
```

```python
# Save the grid directly as a PNG image (no axes, no plot)
print(f"Saving rule {n} with {'random' if r else 'single'} start...")
plt.imsave(f"elementary_ca_rule_150r_{steps}_{cells}{"r" if r else ""}.png", grid, cmap='binary')
```
